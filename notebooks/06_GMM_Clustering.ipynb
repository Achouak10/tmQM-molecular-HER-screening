```python
#Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
```

```python
#Load and Scale Data ---
df = pd.read_csv('../data/tmQM_y.csv')
features = ['HL_Gap', 'Dipole_M', 'Molecular Size', 'MND', 'q']
X = df[features]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Run GMM ---
# Enforced random_state and full covariance for reproducibility
gmm = GaussianMixture(n_components=6, random_state=42, covariance_type='full')
df['GMM_Label'] = gmm.fit_predict(X_scaled)
```

```python
#Export Data ---
df.to_csv('../data/GMM_clustering_results.csv', index=False)
print("Clustering results successfully saved to GMM_clustering_results.csv")
```

```python
#Visualization ---
plt.figure(figsize=(10, 6))

# Map specific colors to the 6 clusters
colors = ['lightgreen', 'darkblue', 'purple', 'yellow', 'green', 'darkred']

for i in range(6):
    subset = df[df['GMM_Label'] == i]
    plt.scatter(subset['MND'], subset['q'], s=15, c=colors[i], label=f'Cluster {i}', alpha=0.6)

plt.tick_params(axis='both', which='major', labelsize=12)
plt.xlabel('Metal Node Degree (MND)', fontsize=14)
plt.ylabel('Total Charge (q)', fontsize=14)
plt.title('GMM Clusters: Coordination vs Total Charge')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
